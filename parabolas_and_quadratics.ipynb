```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(-4, 4, 100)
y = x ** 2
y1 = (x ** 2) + 2
y2 = (-(x - 1) ** 2) + 1
```

```python
plt.plot(x, y, label='y = x^2', color='red')
plt.plot(x, y1, label='y = x^2 + 2')
plt.plot(x, y2, label='y = -(x - 1) ^ 2 + 1')
plt.grid(True)
plt.legend()
plt.show
```

```python
print(x[np.argmax(y2)], y2.max())
```
