<h1 align=center>Notes for <b>‘The quadratic formula’</b></h1>

<h1>Important Ideas and Useful Facts:</h1>
<p>(I) Solving a quadratic equation by inspection: In simple cases, one is able to solve a quadratic
equation by factorising the quadratic by inspection. This technique relies on the fact that
(x + a)(x + b) = x² + (a + b)x + ab .</p>
<p>This method works best when the coefficient of x² is 1. One looks for numbers that
add up to give the coefficient of x and multiply to give the constant term. For example,
consider the quadratic equation x² + 4x − 12 = 0.</p>
<p>Thinking about the integer factorisations of −12, one can see that −12 = (+6) × (−2) 
    and (+6) + (−2) = 4, so the above formula will work with a = 6 and b = −2. 
    Hence the equation becomes (x + 6)(x − 2) = 0, so that either x + 6 = 0 or x − 2 = 0, 
    yielding solutions x = −6 and x = 2.</p>
