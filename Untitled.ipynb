```python
import numpy as np
import pandas as pd
```

```python
data = {
    'Departament': ['IT', 'IT', 'Vânzări', 'Vânzări', 'HR', 'HR'],
    'Angajat': ['Ana', 'Barbu', 'Cristian', 'Diana', 'Elena', 'Florin'],
    'Salariu': [5000, 6000, 4000, 4500, 3500, 3800],
    'Vechime_Ani': [2, 5, 3, 4, 1, 3]
}

df = pd.DataFrame(data)
print(df)
```

```python
df.groupby('Departament')['Salariu'].mean()
```

```python
df.groupby('Departament')['Salariu'].agg(['min', 'max', 'mean', 'median', 'std', 'var'])
```

```python
df['Salariu_mediu'] = df.groupby('Departament')['Salariu'].transform('mean')
```

```python
df['Diferenta'] = df['Salariu_mediu'] - df['Salariu']
```

```python
df
```

```python
df_salarii_mari = df.groupby('Salariu').filter(lambda x: (x['Diferenta'] < 0).all())
```
