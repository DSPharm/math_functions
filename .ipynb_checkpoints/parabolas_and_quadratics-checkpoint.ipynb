```python
import numpy as np
import matplotlib.pyplot as plt
```

**Quadratic expressions:** A quadratic expression (in x) has the form:
                                   ax²+ bx + c
where x is a variable and a, b and c are constants such that a is nonzero. (If a were zero,
then the expression would collapse to bx + c, which is called a linear expression.)

**Parabolas:** A parabola is a curve in the plane that is the result of any combinations of
(positive or negative) dilations, rotations or translations of the curve y = x².
The curve associated with any quadratic expression y = ax² + bx + c is a
parabola, **facing upwards** if a > 0, and **facing downwards** if a < 0.

```python
x = np.linspace(-4, 4, 100)
y = x ** 2
y1 = (x ** 2) + 2
y2 = x ** 2 - 4 * x + 4
```

```python
plt.plot(x, y, label='y = x^2', color='red')
plt.plot(x, y1, label='y = x^2 + 2')
plt.plot(x, y2, label='y = -(x - 1) ^ 2 + 1')
plt.grid(True)
plt.legend()
plt.show
```
