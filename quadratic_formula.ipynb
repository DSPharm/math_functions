<h1 align=center>Notes for <b>‘The quadratic formula’</b></h1>

<h1>Important Ideas and Useful Facts:</h1>
<p>(I) Solving a quadratic equation by inspection: In simple cases, one is able to solve a quadratic
equation by factorising the quadratic by inspection. This technique relies on the fact that
(x + a)(x + b) = x² + (a + b)x + ab .</p>
<p>This method works best when the coefficient of x² is 1. One looks for numbers that
add up to give the coefficient of x and multiply to give the constant term. For example,
consider the quadratic equation x² + 4x − 12 = 0.</p>
<p>Thinking about the integer factorisations of −12, one can see that −12 = (+6) × (−2) 
    and (+6) + (−2) = 4, so the above formula will work with a = 6 and b = −2. 
    Hence the equation becomes (x + 6)(x − 2) = 0, so that either x + 6 = 0 or x − 2 = 0, 
    yielding solutions x = −6 and x = 2.</p>

<p>(II) The quadratic formula: A quadratic equation has the form ax² + bx + c = 0 where a, b,
c are constants, a != 0, and has solutions (also called roots of the quadratic) given by the
quadratic formula:</p>
<p>x =(−b ±√(b² − 4ac))/2a</p>
<p>We call ∆ = b² − 4ac the discriminant. If ∆ > 0 then there are exactly two real roots. 
    If ∆ = 0 then there is exactly one real root.
    If ∆ < 0 then there are no real roots (though exactly two complex roots exist in the 
    complex number system C, though treatments involving complex numbers go beyond the 
    scope of this course).</p>
<p>For example, if a = 1, b = 4 and c = −12 then we are solving the quadratic equation
    x² + 4x − 12 = 0, that we solved earlier, by the factorisation method. If we apply the quadratic formula,
    we get x = (−4 ± √(4² − (4 × 1 × (−12))) / (2 × 1) = (−4 ± √64) / 2 = (−4 ± 8) / 2 = −2 ± 4 ,
    yielding solutions x = 2 and x = −6, as before.</p>
